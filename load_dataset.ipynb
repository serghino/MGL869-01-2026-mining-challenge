# Data Analysis and Visualization Demo

This notebook demonstrates basic usage of pandas, plotly.express, matplotlib, and numpy libraries.

## 1. Import Required Libraries

Import pandas, plotly.express, matplotlib.pyplot, and numpy.

```python
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Create Sample Data with Pandas

Create a DataFrame using pandas with sample data for demonstration purposes.

```python
# Create sample data
data = {
    'Month': ['January', 'February', 'March', 'April', 'May', 'June'],
    'Sales': [15000, 18000, 22000, 19000, 24000, 27000],
    'Expenses': [12000, 13500, 15000, 14000, 16000, 17500],
    'Profit': [3000, 4500, 7000, 5000, 8000, 9500]
}

df = pd.DataFrame(data)
print("Sample DataFrame:")
df
```

## 3. Visualize Data with Plotly Express

Use Plotly Express to create interactive visualizations from the DataFrame.

```python
# Create an interactive line chart with Plotly Express
fig = px.line(df, x='Month', y=['Sales', 'Expenses', 'Profit'], 
              title='Monthly Sales, Expenses, and Profit',
              labels={'value': 'Amount ($)', 'variable': 'Category'})
fig.show()
```

```python
# Create an interactive bar chart
fig2 = px.bar(df, x='Month', y='Profit', 
              title='Monthly Profit',
              color='Profit',
              color_continuous_scale='viridis')
fig2.show()
```

## 4. Create Plots with Matplotlib

Use Matplotlib to create static plots and customize their appearance.

```python
# Create a figure with multiple subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Line chart
axes[0].plot(df['Month'], df['Sales'], marker='o', label='Sales', linewidth=2)
axes[0].plot(df['Month'], df['Expenses'], marker='s', label='Expenses', linewidth=2)
axes[0].set_title('Sales vs Expenses')
axes[0].set_xlabel('Month')
axes[0].set_ylabel('Amount ($)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)
axes[0].tick_params(axis='x', rotation=45)

# Plot 2: Bar chart
axes[1].bar(df['Month'], df['Profit'], color='green', alpha=0.7)
axes[1].set_title('Monthly Profit')
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Profit ($)')
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```

## 5. Perform Numerical Operations with NumPy

Demonstrate basic NumPy array operations and mathematical computations.

```python
# Convert DataFrame columns to NumPy arrays
sales_array = np.array(df['Sales'])
expenses_array = np.array(df['Expenses'])

print("Sales Array:", sales_array)
print("Expenses Array:", expenses_array)
print()

# Basic statistics
print("Sales Statistics:")
print(f"Mean: ${np.mean(sales_array):,.2f}")
print(f"Median: ${np.median(sales_array):,.2f}")
print(f"Standard Deviation: ${np.std(sales_array):,.2f}")
print(f"Min: ${np.min(sales_array):,.2f}")
print(f"Max: ${np.max(sales_array):,.2f}")
```

```python
# Perform mathematical operations
profit_margin = ((sales_array - expenses_array) / sales_array) * 100
cumulative_sales = np.cumsum(sales_array)

print("Profit Margin (%):", np.round(profit_margin, 2))
print("Cumulative Sales:", cumulative_sales)
print()

# Matrix operations
data_matrix = np.column_stack((sales_array, expenses_array))
print("Data Matrix:")
print(data_matrix)
print()
print("Matrix Transpose:")
print(data_matrix.T)
```

```python
# Generate random data for demonstration
random_data = np.random.normal(loc=20000, scale=3000, size=100)
print(f"Generated {len(random_data)} random samples")
print(f"Mean: ${np.mean(random_data):,.2f}")
print(f"Std Dev: ${np.std(random_data):,.2f}")

# Create histogram
plt.figure(figsize=(10, 5))
plt.hist(random_data, bins=20, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(np.mean(random_data), color='red', linestyle='--', linewidth=2, label='Mean')
plt.title('Distribution of Random Sales Data')
plt.xlabel('Sales Amount ($)')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
